# Earthquake Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from matplotlib.colors import LinearSegmentedColormap

barbie_pink = LinearSegmentedColormap.from_list(
    "barbie_pink",
    ["#ffe6f2", "#ffb3d9", "#ff80bf", "#ff4da6", "#ff1a8c"]
)

magenta_glow = LinearSegmentedColormap.from_list(
    "magenta_glow",
    ["#fff0f6", "#ffb3e6", "#ff66d9", "#cc33b0", "#990f80"]
)

danger = LinearSegmentedColormap.from_list(
    "danger",
    ["#ffcccc", "#ff6666", "#ff0000", "#990000", "#000000"]
)
```

```python
quakes = pd.read_csv('data/quake.csv')
```

```python
quakes = quakes.drop_duplicates()
```

```python
plt.figure(figsize=(12,8))

plt.scatter(
    x=quakes['longitude'],
    y=quakes['latitude'],
    alpha=1,
    c=quakes['richter'],
    cmap=danger
)
plt.colorbar(label='Richter Scale')
plt.grid(color="#cccccc", linestyle="-", linewidth=0.4, alpha=0.4)
plt.title('Earthquakes by Richter Scale')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.show()
```
